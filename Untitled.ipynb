 Python Collections Module
Python's collections module provides specialized container data types that extend the functionality of Python's built-in data structures like list, tuple, dict, and set.

It is particularly useful in data science, algorithms, and efficient data manipulation.

First, import the module:

```python
from collections import Counter, defaultdict, deque, namedtuple, ChainMap
```

Let's understand each one with examples.
1. Counter
Very important
Counter is a dictionary subclass used to count the frequency of elements in an iterable.

Example 1: Counting characters

```python
from collections import Counter

word = "banana"

result = Counter(word)

print(result)
```

```python
from collections import Counter

words = ["apple", "banana", "apple",
         "orange", "banana", "apple"]

count = Counter(words)

print(count)
print(count["apple"])
print(count.most_common(2))
```

Normally, accessing a nonexistent dictionary key produces a KeyError.

defaultdict automatically creates a default value for missing keys.

Normal dictionary:

```python
students = {}

students["Ram"].append(90)  # KeyError
```

```python
from collections import defaultdict

students = defaultdict(list)

students["Ram"].append(90)
students["Ram"].append(85)

students["Sita"].append(95)

print(dict(students))
```

You can use different default factories:

```python
defaultdict(list)   # Default value: []
defaultdict(int)    # Default value: 0
defaultdict(set)    # Default value: set()
defaultdict(float)  # Default value: 0.0
```

Example: Grouping students by department

```python
from collections import defaultdict

students = [
    ("Ram", "Computer"),
    ("Sita", "Civil"),
    ("Hari", "Computer"),
    ("Gita", "Civil")
]

departments = defaultdict(list)

for name, department in students:
    departments[department].append(name)

print(dict(departments))
```

3. deque (Double-Ended Queue)
Important for DSA
A deque allows insertion and deletion from both ends efficiently.

Think of it as a queue where you can work from both the front and the back.

```python
from collections import deque

numbers = deque([10, 20, 30])

numbers.append(40)       # Add to right
numbers.appendleft(5)    # Add to left

print(numbers)

numbers.pop()            # Remove from right
numbers.popleft()        # Remove from left

print(numbers)
```

```python
from collections import deque

d = deque([1, 2, 3, 4])

d.rotate(1)

print(d)
```

```python
from collections import namedtuple

Student = namedtuple("Student", ["name", "age", "department"])

s1 = Student("Ram", 20, "Computer")

print(s1.name)
print(s1.age)
print(s1.department)
```

5. ChainMap
ChainMap combines multiple dictionaries into a single view without merging them into a new dictionary.

```python
from collections import ChainMap

default_settings = {
    "theme": "light",
    "language": "English"
}

user_settings = {
    "theme": "dark"
}

settings = ChainMap(user_settings, default_settings)

print(settings["theme"])
print(settings["language"])
```

Why?
- First, Python searches user_settings.
- If the key is missing, it searches default_settings.
This is useful for configuration management.

6. OrderedDict
OrderedDict is a dictionary subclass that provides explicit ordering operations.

Modern Python dictionaries already preserve insertion order, so OrderedDict is less necessary than it used to be.

However, it provides additional methods.

```python
from collections import OrderedDict

data = OrderedDict()

data["a"] = 10
data["b"] = 20
data["c"] = 30

data.move_to_end("a")

print(data)
```

## Iterators

An iterator is an object that allows us to access elements of a collection (list, tuple, dictionary, etc.) one at a time.

Instead of accessing all elements simultaneously, an iterator remembers its current position and gives us the next element whenever requested.

Think of reading a book:
- You start at page 1.
- You read one page.
- You move to page 2.
- You continue until the book ends.
An iterator works similarly!
1. Iterable vs Iterator
These two terms are very important.

Iterable: An object that can be iterated over (traversed).

Examples:
- List
- Tuple
- String
- Dictionary
- Set
Iterator: An object that actually produces elements one at a time.

1. Two important functions
- iter() – Converts an iterable into an iterator.
- next() – Returns the next element.

```python
numbers = [10, 20, 30]

it = iter(numbers)

print(next(it))  # 10
print(next(it))  # 20
print(next(it))  # 30
```

2. StopIteration
When no elements remain, Python raises StopIteration.

```python
numbers = [10, 20]
it = iter(numbers)

print(next(it))
print(next(it))
print(next(it))  # StopIteration
```

```python
class MyNumbers:

    def __init__(self):
        self.x = 1

    def __iter__(self):
        return self

    def __next__(self):
        if self.x <= 3:
            value = self.x
            self.x += 1
            return value
        raise StopIteration

obj = MyNumbers()

for i in obj:
    print(i)
```

5. Advantages
1. Memory efficient.
2. Processes large datasets one element at a time.
3. Useful for file handling and data processing.

Python Generators
Definition: A generator is a special type of iterator that generates values one at a time using the yield keyword.

Example:

```python
def numbers():
    yield 10
    yield 20
    yield 30

for n in numbers():
    print(n)
```

Important points:
- yield returns a value without terminating the function.
- It remembers its previous state.
- Generates values one at a time.
- Memory efficient for large datasets.
Generator vs Normal Function:
    Normal function	Generator
Uses return	Uses yield
Returns a result	Produces values one at a time
Function terminates at return	Pauses and resumes at yield

Remember: Generators are useful for memory-efficient processing of large amounts of data.

Python Decorators
Definition: A decorator is a function that modifies or extends the behavior of another function without changing its original code.

Example:

```python
def my_decorator(func):
    def wrapper():
        print("Before function")
        func()
        print("After function")
    return wrapper

@my_decorator
def greet():
    print("Hello!")

greet()
```

Important points:
- Decorators use the @ symbol.
- They accept another function as an argument.
- They add functionality without modifying the original function.
- Common uses: logging, authentication, timing and debugging.
Remember: @my_decorator is equivalent to greet = my_decorator(greet).

## OOP for data science


OOP (Object-Oriented Programming) for Data Science
Definition: OOP is a programming approach that organizes code into classes and objects, making data and related functions easier to manage and reuse.

1. Class and Object
Class: Blueprint for creating objects.

Object: Instance of a class.

```python
class Student:
    def __init__(self, name, marks):
        self.name = name
        self.marks = marks

    def display(self):
        print(self.name, self.marks)

s1 = Student("Ram", 90)
s1.display()
```

2. Four pillars of OOP
Concept	Meaning
Encapsulation	Combining data and methods in a class
Inheritance	Reusing properties of another class
Polymorphism	Same method, different behavior
Abstraction	Hiding implementation details

3. Application in Data Science
OOP helps organize data science projects.

Example: Student Data Analysis

```python
class DataAnalyzer:

    def __init__(self, data):
        self.data = data

    def average(self):
        return sum(self.data) / len(self.data)

    def maximum(self):
        return max(self.data)


marks = DataAnalyzer([80, 90, 70, 85])

print(marks.average())
print(marks.maximum())
```

Practical OOP Example: Employee Salary Analysis System
Imagine you work as a Data Scientist in a company. Your manager gives you employee salary data and asks you to calculate average salaries, bonuses, and employee performance.

We'll use all four pillars of OOP.

```python
# 1. ENCAPSULATION
class Employee:

    def __init__(self, name, salary, performance):
        self.name = name
        self.__salary = salary  # Private attribute
        self.performance = performance

    def get_salary(self):
        return self.__salary

    def give_bonus(self):
        self.__salary += 5000
        return self.__salary


# 2. INHERITANCE
class DataScientist(Employee):

    def analyze(self):
        return "Analyzing company data"


# 3. POLYMORPHISM
class Manager(Employee):

    def analyze(self):
        return "Analyzing team performance"


# 4. ABSTRACTION
from abc import ABC, abstractmethod

class Report(ABC):

    @abstractmethod
    def generate(self):
        pass


class SalaryReport(Report):

    def generate(self):
        return "Generating employee salary report"


# Creating objects
e1 = DataScientist("Ram", 50000, 90)
e2 = Manager("Sita", 80000, 95)

# Data analysis
employees = [e1, e2]

salaries = [e.get_salary() for e in employees]

print("Average Salary:", sum(salaries)/len(salaries))

# Polymorphism
for e in employees:
    print(e.name, e.analyze())

# Bonus
print("Ram's salary after bonus:", e1.give_bonus())

# Abstraction
report = SalaryReport()
print(report.generate())
```

How the four pillars are used
Pillar	Application
Encapsulation	Salary is private and accessed through methods.
Inheritance	DataScientist and Manager inherit Employee.
Polymorphism	Both classes implement analyze() differently.
Abstraction	Report defines a required generate() method without implementing it.

Real-world office applications
- Encapsulation: Protect sensitive employee salary data.
- Inheritance: Reuse common employee properties.
- Polymorphism: Generate different analyses for different employee roles.
- Abstraction: Create a common structure for different report types.
Key takeaway: OOP helps data scientists build reusable, organized and maintainable data analysis systems.

## Exception handling, debugging, and logging 

1. Exception Handling
Definition: Exception handling is a mechanism used to handle runtime errors without stopping the entire program.

Main keywords:
- try: Contains code that may cause an error.
- except: Handles the error.
- else: Executes if no exception occurs.
- finally: Always executes.
- raise: Manually generates an exception.

```python
try:
    a = 10 / 0
except ZeroDivisionError:
    print("Cannot divide by zero")
else:
    print("No error occurred")
finally:
    print("Program completed")
```

Example: Employee Salary Analysis
Imagine a company has employee salary data. We need to calculate the average salary, handle invalid data, debug problems, and record program activities.

```python

import logging

# 1. LOGGING CONFIGURATION
logging.basicConfig(
    filename="company.log",
    level=logging.DEBUG,
    format="%(levelname)s: %(message)s"
)

# Employee salary data
employees = {
    "Ram": 50000,
    "Sita": 60000,
    "Hari": 45000,
    "Gita": "invalid"
}

# 2. FUNCTION TO CALCULATE AVERAGE SALARY
def calculate_average(data):

    total = 0
    count = 0

    logging.info("Salary calculation started")

    for name, salary in data.items():

        try:
            # Convert salary into a number
            salary = float(salary)

            # DEBUGGING: Check values during execution
            logging.debug(f"Processing {name}: {salary}")

            if salary < 0:
                raise ValueError("Salary cannot be negative")

            total += salary
            count += 1

        except ValueError as error:
            logging.error(f"Invalid salary for {name}: {error}")
            print(f"Skipping {name}: Invalid salary")

        except TypeError as error:
            logging.error(f"Type error: {error}")

    # Avoid division by zero
    try:
        average = total / count

    except ZeroDivisionError:
        logging.critical("No valid salary records found")
        return 0

    else:
        logging.info("Average calculated successfully")
        return average

    finally:
        logging.info("Salary calculation completed")


# 3. MAIN PROGRAM
try:
    result = calculate_average(employees)
    print("Average Salary:", result)

except Exception as error:
    logging.exception("Unexpected program error")
    print("Something went wrong:", error)
```

```python
import csv

# STEP 1: CREATE CSV FILE
employees = [
    ["Name", "Age", "Salary"],
    ["Ram", 22, 50000],
    ["Sita", 24, 60000],
    ["Hari", 23, 45000]
]

with open("employees.csv", "w", newline="") as file:
    writer = csv.writer(file)
    writer.writerows(employees)

print("CSV file created successfully!")


# STEP 2: READ CSV FILE
with open("employees.csv", "r") as file:
    reader = csv.DictReader(file)
    data = list(reader)

print("\nOriginal Dataset:")
for employee in data:
    print(employee)


# STEP 3: MODIFY DATA - UPDATE SALARY
for employee in data:
    if employee["Name"] == "Ram":
        employee["Salary"] = "55000"


# STEP 4: ADD A NEW EMPLOYEE
data.append({
    "Name": "Gita",
    "Age": "25",
    "Salary": "65000"
})


# STEP 5: DELETE AN EMPLOYEE
data = [employee for employee in data
        if employee["Name"] != "Hari"]


# STEP 6: WRITE MODIFIED DATA BACK TO CSV
with open("employees.csv", "w", newline="") as file:
    writer = csv.DictWriter(
        file,
        fieldnames=["Name", "Age", "Salary"]
    )

    writer.writeheader()
    writer.writerows(data)


# STEP 7: PRINT COMPLETE FINAL DATASET
print("\nFinal CSV Dataset:")

with open("employees.csv", "r") as file:
    reader = csv.DictReader(file)

    for employee in reader:
        print(employee)
```

### 2.2 Database Access: Relational and Non-Relational Databases

```python
import sqlite3

# STEP 1: CONNECT TO DATABASE
# If the database does not exist, SQLite creates it.
connection = sqlite3.connect("company.db")

# Create a cursor to execute SQL commands
cursor = connection.cursor()


# STEP 2: CREATE TABLE
cursor.execute("""
CREATE TABLE IF NOT EXISTS employees (
    id INTEGER PRIMARY KEY,
    name TEXT,
    age INTEGER,
    salary REAL
)
""")


# STEP 3: INSERT DATA
employees = [
    (1, "Ram", 22, 50000),
    (2, "Sita", 24, 60000),
    (3, "Hari", 23, 45000)
]

cursor.executemany("""
INSERT OR IGNORE INTO employees
(id, name, age, salary)
VALUES (?, ?, ?, ?)
""", employees)

connection.commit()


# STEP 4: READ DATA
print("Original Dataset:")

cursor.execute("SELECT * FROM employees")

for row in cursor.fetchall():
    print(row)


# STEP 5: UPDATE DATA
# Change Ram's salary to 55000
cursor.execute("""
UPDATE employees
SET salary = 55000
WHERE name = 'Ram'
""")


# STEP 6: DELETE DATA
# Remove Hari's record
cursor.execute("""
DELETE FROM employees
WHERE name = 'Hari'
""")


# STEP 7: INSERT A NEW RECORD
cursor.execute("""
INSERT OR IGNORE INTO employees
(id, name, age, salary)
VALUES (4, 'Gita', 25, 65000)
""")

connection.commit()


# STEP 8: PRINT FINAL DATASET
print("\nFinal Dataset:")

cursor.execute("SELECT * FROM employees")

for row in cursor.fetchall():
    print(row)


# STEP 9: CLOSE DATABASE CONNECTION
connection.close()
```

Accessing SOAP APIs using Python
We can use the requests library to send XML messages.

```python
import requests

# SOAP service URL
url = "https://example.com/student-service"

# SOAP XML request
xml_request = """
<soap:Envelope
xmlns:soap="http://schemas.xmlsoap.org/soap/envelope/">

    <soap:Body>
        <GetStudent>
            <StudentID>101</StudentID>
        </GetStudent>
    </soap:Body>

</soap:Envelope>
"""

# Send SOAP request
response = requests.post(
    url,
    data=xml_request,
    headers={
        "Content-Type": "text/xml; charset=utf-8"
    }
)

# Display response
print(response.status_code)
print(response.text)
```

```python
import requests

url = "https://www.dataaccess.com/webservicesserver/NumberConversion.wso"

xml_data = """<?xml version="1.0" encoding="utf-8"?>
<soap:Envelope xmlns:soap="http://schemas.xmlsoap.org/soap/envelope/">
    <soap:Body>
        <NumberToWords xmlns="http://www.dataaccess.com/webservicesserver/">
            <ubiNum>125</ubiNum>
        </NumberToWords>
    </soap:Body>
</soap:Envelope>
"""

headers = {
    "Content-Type": "text/xml; charset=utf-8",
    "SOAPAction": '""',
    "User-Agent": "Mozilla/5.0",
    "Accept": "text/xml"
}

response = requests.post(
    url,
    data=xml_data,
    headers=headers,
    timeout=20
)

print("Status:", response.status_code)
print(response.text)
```

```python
import xml.etree.ElementTree as ET

# Store the SOAP response
xml_response = response.text

# Parse XML into a tree
root = ET.fromstring(xml_response)

# Define XML namespaces
namespaces = {
    "soap": "http://schemas.xmlsoap.org/soap/envelope/",
    "m": "http://www.dataaccess.com/webservicesserver/"
}

# Find the result element
result = root.find(
    ".//m:NumberToWordsResult",
    namespaces
)

# Print extracted result
print("Number in words:", result.text.strip())
```

```python
import requests
from bs4 import BeautifulSoup
import pandas as pd

# Website URL
url = "https://books.toscrape.com/"

# Send HTTP GET request
response = requests.get(url, timeout=20)

# Check whether the request was successful
response.raise_for_status()

# Parse HTML content
soup = BeautifulSoup(response.text, "html.parser")

# Find all book containers
books = soup.find_all("article", class_="product_pod")

# Create an empty list to store book information
book_data = []

# Extract information from each book
for book in books:

    # Extract book title from the link's title attribute
    title = book.find("h3").find("a").get("title")

    # Extract book price
    price = book.find("p", class_="price_color").get_text(strip=True)

    # Extract availability
    availability = book.find(
        "p", class_="instock availability"
    ).get_text(strip=True)

    # Extract rating from the CSS class
    rating = book.find("p", class_="star-rating")["class"][1]

    # Store information in a dictionary
    book_data.append({
        "Title": title,
        "Price": price,
        "Availability": availability,
        "Rating": rating
    })

# Convert list of dictionaries into DataFrame
df = pd.DataFrame(book_data)

# Display complete dataset
print(df.to_string(index=False))

# Save extracted data into CSV file
df.to_csv("books_data.csv", index=False)

print("\nData successfully saved!")
```

```python
import pandas as pd

# Initialize total salary
total_salary = 0

# Initialize employee count
total_employees = 0

# Read file in chunks of 1000 rows
for chunk in pd.read_csv("employees.csv", chunksize=1000):

    # Add salary of current chunk
    total_salary += chunk["Salary"].sum()

    # Count employees in current chunk
    total_employees += len(chunk)

# Calculate overall average
average_salary = total_salary / total_employees

print("Total employees:", total_employees)
print("Total salary:", total_salary)
print("Average salary:", average_salary)
```

## Automating workflows with schedulers (CRON, schedule)

```python
%pip install schedule
```

```python
import pandas as pd
import schedule
import time
from datetime import datetime

# Define the data processing task
def generate_report():

    print("Starting data processing...")

    # Read employee dataset
    df = pd.read_csv("employees.csv")

    # Calculate average salary
    average_salary = df["Salary"].mean()

    # Calculate total employees
    total_employees = len(df)

    # Generate report
    report = f"""
    EMPLOYEE REPORT
    -------------------------
    Date: {datetime.now()}
    Total Employees: {total_employees}
    Average Salary: {average_salary}
    """

    # Save report to a text file
    with open("report.txt", "w") as file:
        file.write(report)

    print("Report generated successfully!")

# Schedule task every 10 seconds
schedule.every(10).seconds.do(generate_report)

# Keep scheduler running
while True:

    # Execute pending tasks
    schedule.run_pending()

    # Wait before checking again
    time.sleep(1)
```
